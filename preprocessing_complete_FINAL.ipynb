<h3 align="left">
<font color="##167D92">1. Content-Based Recommendation
</font>
</h3>

</div>

Content-based recommendation suggests products that are similar
to products a user has previously interacted with.

The similarity is determined from the characteristics or content
of the products.

For textual product information, TF-IDF can be used to transform
product descriptions into numerical vectors. Cosine similarity
can then be used to measure similarity between products.

***"You liked this product, so I'll find products that look similar to it."***

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="##167D92">2. Collaborative Filtering
</font>
</h3>

</div>



Collaborative filtering generates recommendations based on
user-product interactions.

Instead of focusing primarily on the characteristics of a product,
the method learns from patterns in user behavior.

For example, users with similar preferences may receive similar
recommendations.

***"People who behave like you liked these products, so you may like them too."***

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="##167D92">3. Hybrid Recommendation
</font>
</h3>

</div>


A hybrid recommendation system combines content-based and
collaborative filtering approaches.

The objective is to benefit from both product characteristics
and user interaction patterns.

***"The product is similar to things you like AND people with similar tastes also liked it."***

<h1 align="left"><font color=#ff6200>06 — Content-Based Filtering</font></h1>

</div>

This recommender suggests products that are **similar to a given product**,
based on the text of their `Description`. The idea: *"You liked this
product, here are others like it."*

Approach:
1. Build a table of unique products (one row per `StockCode`).
2. Vectorize each product's description with **TF-IDF**.
3. Compute **cosine similarity** between all products.
4. For a given product, return the top-N most similar ones.



<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Building the Unique Product Table</font>
</h3>

</div>

The same `StockCode` can repeat across many transactions — we only need
one row per product to compute similarity.


```python
products = (
    df_model[['StockCode', 'Description']]
    .drop_duplicates(subset='StockCode')
    .reset_index(drop=True)
)

print(f"Unique products: {products.shape[0]}")
products.head()
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">TF-IDF Vectorization</font>
</h3>

</div>

TF-IDF turns each `Description` into a vector where common words across
all products (e.g. "SET", "BAG") weigh less, and words that are
distinctive to a product weigh more.


```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(products['Description'])

print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Cosine Similarity Matrix</font>
</h3>

</div>

Cosine similarity measures how close two product descriptions are in
the TF-IDF vector space (1 = identical, 0 = nothing in common).


```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# Map StockCode -> row index in the similarity matrix, for fast lookup
product_index = pd.Series(products.index, index=products['StockCode']).drop_duplicates()

cosine_sim.shape
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Recommendation Function</font>
</h3>

</div>


```python
def recommend_content_based(stock_code, top_n=5):
    """Return the top_n products most similar to the given StockCode."""
    if stock_code not in product_index:
        print(f"StockCode '{stock_code}' not found.")
        return None

    idx = product_index[stock_code]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Skip index 0 since it's the product itself
    sim_scores = sim_scores[1:top_n + 1]
    product_indices = [i for i, score in sim_scores]
    scores = [round(score, 3) for i, score in sim_scores]

    result = products.iloc[product_indices][['StockCode', 'Description']].copy()
    result['SimilarityScore'] = scores
    return result.reset_index(drop=True)
```

<font color="#0E4F70">Testing the Recommender</font>
</h3>

</div>



```python
sample_code = products['StockCode'].iloc[0]
print("Product:", products[products['StockCode'] == sample_code]['Description'].values[0])
print()
recommend_content_based(sample_code, top_n=5)
```

<h1 align="left"><font color=#ff6200>07 — Collaborative Filtering</font></h1>

</div>

Unlike content-based filtering (which looks at *what a product is*),
collaborative filtering looks at *who bought what*. The idea: *"Customers
who bought similar products to you also bought this."*

We use an **item-based** approach here:
1. Build a Customer × Product interaction matrix (quantities purchased).
2. Compute similarity **between products**, based on co-purchase patterns
   across all customers (not their descriptions).
3. For a given customer, recommend products similar to what they already
   bought — excluding items they've already purchased.

Item-based CF is generally more stable than user-based CF for e-commerce,
since product relationships change far less often than individual
customer taste.


<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Building the Customer-Product Interaction Matrix</font>
</h3>

</div>

Rows = customers, columns = products, values = total quantity purchased.
Most cells are 0 (a customer only buys a small fraction of the catalog),
so this matrix is naturally sparse.


```python
interaction_matrix = df_model.pivot_table(
    index='CustomerID',
    columns='StockCode',
    values='Quantity',
    aggfunc='sum',
    fill_value=0
)

print("Interaction matrix shape (customers x products):", interaction_matrix.shape)
interaction_matrix.iloc[:5, :5]
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Item-Item Similarity Matrix</font>
</h3>

</div>

We transpose the matrix so that rows become products, then compute
cosine similarity between products based on their purchase patterns
across all customers. We use a sparse matrix here since most values
are 0 — this keeps the computation efficient on large datasets.



```python
from scipy.sparse import csr_matrix
from sklearn.metrics.pairwise import cosine_similarity

item_matrix_sparse = csr_matrix(interaction_matrix.T.values)
item_sim_cf = cosine_similarity(item_matrix_sparse)

# Map StockCode -> row index in this similarity matrix
cf_item_index = pd.Series(
    range(len(interaction_matrix.columns)),
    index=interaction_matrix.columns
)

item_sim_cf.shape
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Recommendation Function</font>
</h3>

</div>


```python
def recommend_collaborative(customer_id, top_n=5):
    """Recommend products for a customer based on item-item collaborative filtering."""
    if customer_id not in interaction_matrix.index:
        print(f"CustomerID '{customer_id}' not found.")
        return None

    # Products this customer already bought
    customer_row = interaction_matrix.loc[customer_id]
    purchased_items = customer_row[customer_row > 0].index.tolist()

    if not purchased_items:
        print(f"CustomerID '{customer_id}' has no purchase history.")
        return None

    # Aggregate similarity scores across all items the customer bought
    scores = pd.Series(0, index=interaction_matrix.columns, dtype=float)
    for item in purchased_items:
        idx = cf_item_index[item]
        scores += item_sim_cf[idx]

    # Remove items already purchased
    scores = scores.drop(index=purchased_items)

    top_items = scores.sort_values(ascending=False).head(top_n)

    result = products[products['StockCode'].isin(top_items.index)].copy()
    result['Score'] = result['StockCode'].map(top_items).round(3)
    return result.sort_values('Score', ascending=False).reset_index(drop=True)
```

</h3>

</div>

<h1 align="left"><font color=#ff6200>08 — Hybrid Recommender</font></h1>

</div>

Content-based and collaborative filtering each have blind spots:
- **Content-based** ignores purchase behavior — it can't tell that two
  differently-described products are often bought together.
- **Collaborative filtering** ignores product descriptions — it struggles
  with new/rarely-purchased products (the *cold-start* problem).

The **hybrid** approach combines both similarity scores into one, using
a weight `alpha` to control how much each method contributes:

`hybrid_score = alpha * content_based_score + (1 - alpha) * collaborative_score`


<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Combining the Two Similarity Scores</font>
</h3>

</div>

Both similarity matrices (`cosine_sim` from content-based, `item_sim_cf`
from collaborative) are on different scales, so we normalize each to a
0–1 range before combining them.


```python
def get_similarity_scores(purchased_items, sim_matrix, index_map, all_items):
    """Aggregate similarity scores across a list of purchased items."""
    scores = pd.Series(0, index=all_items, dtype=float)
    for item in purchased_items:
        if item in index_map:
            idx = index_map[item]
            scores += sim_matrix[idx]
    return scores


def normalize(scores):
    """Min-max normalize a score series to the 0-1 range."""
    if scores.max() == scores.min():
        return scores * 0
    return (scores - scores.min()) / (scores.max() - scores.min())
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Hybrid Recommendation Function</font>
</h3>

</div>


```python
def recommend_hybrid(customer_id, top_n=5, alpha=0.5):
    """
    Recommend products for a customer combining content-based and
    collaborative filtering scores.

    alpha : weight given to content-based similarity (0 to 1).
            (1 - alpha) is the weight given to collaborative similarity.
    """
    if customer_id not in interaction_matrix.index:
        print(f"CustomerID '{customer_id}' not found.")
        return None

    customer_row = interaction_matrix.loc[customer_id]
    purchased_items = customer_row[customer_row > 0].index.tolist()

    if not purchased_items:
        print(f"CustomerID '{customer_id}' has no purchase history.")
        return None

    common_items = interaction_matrix.columns.intersection(products['StockCode'])

    cf_scores = get_similarity_scores(purchased_items, item_sim_cf, cf_item_index, interaction_matrix.columns)
    cb_scores = get_similarity_scores(purchased_items, cosine_sim, product_index, products['StockCode'])

    cf_scores = normalize(cf_scores.reindex(common_items).fillna(0))
    cb_scores = normalize(cb_scores.reindex(common_items).fillna(0))

    hybrid_scores = alpha * cb_scores + (1 - alpha) * cf_scores
    hybrid_scores = hybrid_scores.drop(index=[i for i in purchased_items if i in hybrid_scores.index])

    top_items = hybrid_scores.sort_values(ascending=False).head(top_n)

    result = products[products['StockCode'].isin(top_items.index)].copy()
    result['HybridScore'] = result['StockCode'].map(top_items).round(3)
    return result.sort_values('HybridScore', ascending=False).reset_index(drop=True)
```

<font color="#0E4F70">Testing the Hybrid Recommender</font>
</h3>

```python
sample_customer = interaction_matrix.index[0]

print("Balanced hybrid (alpha=0.5):")
display(recommend_hybrid(sample_customer, top_n=5, alpha=0.5))

print("\nContent-leaning hybrid (alpha=0.8):")
display(recommend_hybrid(sample_customer, top_n=5, alpha=0.8))
```

<h1 align="left"><font color=#ff6200>09 — Evaluation & Comparison</font></h1>

</div>

To compare Content-Based, Collaborative, and Hybrid recommendations
objectively, we use a **held-out evaluation**:

1. For each customer with enough purchase history, hide a portion of
   their purchased products (the "test" items).
2. Generate top-K recommendations using only the remaining ("train")
   items.
3. Check how many of the hidden test items appear in the top-K list —
   this gives **Precision@K** and **Recall@K**.

**Note / limitation:** the similarity matrices (`cosine_sim`,
`item_sim_cf`) were computed on the *full* dataset, including the
test interactions. This is a simplification for this stage of the
project — a fully rigorous evaluation would rebuild both similarity
matrices using train-only data, which avoids any information leakage.


<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Train / Test Split (Per Customer)</font>
</h3>

</div>


```python
def train_test_split_customers(interaction_matrix, test_size=0.2, min_items=5, random_state=42):
    """For each customer with enough purchases, hold out a fraction as test items."""
    rng = np.random.default_rng(random_state)
    train_items, test_items = {}, {}

    for customer in interaction_matrix.index:
        purchased = interaction_matrix.loc[customer]
        purchased = purchased[purchased > 0].index.tolist()
        if len(purchased) < min_items:
            continue

        n_test = max(1, int(len(purchased) * test_size))
        test_set = list(rng.choice(purchased, size=n_test, replace=False))
        train_set = [p for p in purchased if p not in test_set]

        train_items[customer] = train_set
        test_items[customer] = test_set

    return train_items, test_items


train_items, test_items = train_test_split_customers(interaction_matrix, test_size=0.2, min_items=5)
print(f"Customers used for evaluation: {len(train_items)}")
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Precision@K and Recall@K</font>
</h3>

</div>


```python
def get_top_k_items(train_list, method='hybrid', alpha=0.5, k=10):
    common_items = interaction_matrix.columns.intersection(products['StockCode'])

    if method == 'cf':
        scores = get_similarity_scores(train_list, item_sim_cf, cf_item_index, interaction_matrix.columns)
    elif method == 'cb':
        scores = get_similarity_scores(train_list, cosine_sim, product_index, products['StockCode'])
    elif method == 'hybrid':
        cf_scores = normalize(get_similarity_scores(train_list, item_sim_cf, cf_item_index, interaction_matrix.columns).reindex(common_items).fillna(0))
        cb_scores = normalize(get_similarity_scores(train_list, cosine_sim, product_index, products['StockCode']).reindex(common_items).fillna(0))
        scores = alpha * cb_scores + (1 - alpha) * cf_scores
    else:
        raise ValueError("method must be 'cf', 'cb', or 'hybrid'")

    scores = scores.drop(index=[i for i in train_list if i in scores.index])
    return scores.sort_values(ascending=False).head(k).index.tolist()


def precision_recall_at_k(train_items, test_items, method='hybrid', alpha=0.5, k=10):
    precisions, recalls = [], []

    for customer, train_list in train_items.items():
        test_list = test_items[customer]
        top_k = get_top_k_items(train_list, method=method, alpha=alpha, k=k)

        hits = len(set(top_k) & set(test_list))
        precisions.append(hits / k)
        recalls.append(hits / len(test_list))

    return np.mean(precisions), np.mean(recalls)
```

<div style="border-radius:10px; padding: 15px; background-color: #EAF4F7; font-size:130%; text-align:left">

<h3 align="left">
<font color="#0E4F70">Comparing the Three Methods</font>
</h3>

</div>

Note: this cell can take a while to run since it scores every eligible
customer for each method — consider evaluating on a sample of
customers first if it's too slow on the full dataset.


```python
K = 10
results = []

for method in ['cb', 'cf', 'hybrid']:
    precision, recall = precision_recall_at_k(train_items, test_items, method=method, alpha=0.5, k=K)
    results.append({'Method': method.upper(), f'Precision@{K}': precision, f'Recall@{K}': recall})

results_df = pd.DataFrame(results)
results_df
```

This project built and compared three approaches to product
recommendation on e-commerce transaction data:

- *Content-Based Filtering* recommends products with similar

  can't capture purchase behavior — two products described very

  patterns across customers, using an item-item similarity matrix. It
  captures real buying behavior, but struggles with new products that
  have little or no purchase history (cold-start problem).

- *Hybrid Recommender* combines both scores with a tunable weight
  (alpha), balancing the strengths of each method.
